```python
import numpy as np
import matplotlib.pyplot as plt

#creating x and y axis
x_train = np.array([2, 4, 6, 8, 10]) 
y_train = np.array([40, 55, 70, 85, 95]) 

#finding m (number of total tarining examples)
m = x_train.shape[0]
print(f"Total number of training examples are {m}")

#get a particualr trainoing example out of all 
i=0
x_i = x_train[i]
y_i = y_train[i]
print(f'x^({i}):{x_i}    y^({i}):{y_i}')

#plot graph
plt.scatter(x_train,y_train,marker='x',c='r')
plt.title('Graph')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()   

#model parameters (weight & bias) 
w=7
b=25
print(f'w:{w}')
print(f'b:{b}')

#linear regression model
def prediction(w,b,x):
    m=x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = x[i]*w+b
    return f_wb
        
#create predictions
fun = prediction(w,b,x_train)

#plot final graph
plt.plot(x_train,fun,c='b',label='prediction')
plt.scatter(x_train,y_train,marker='x',c='r',label='actual values')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Final Prediction Graph')
plt.legend()
plt.show()
```
