```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=2)
```

```python
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])
```

```python
#Sigmoid
def sigmoid(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
# -----------------------------
# Compute Cost
# -----------------------------
def compute_cost(X, y, w, b):

    m = X.shape[0]
    cost = 0.0

    for i in range(m):
        z_i = np.dot(X[i], w) + b
        f_wb_i = sigmoid(z_i)

        loss = -y[i] * np.log(f_wb_i) -((1 - y[i]) * np.log(1 - f_wb_i))

        cost += loss

    cost = cost / m
    return cost
```

```python
# -----------------------------
# Compute Gradient
# -----------------------------
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    n = x.shape[1]
    gradient_w = np.zeros(n)
    gradient_b = 0
    for i in range(m):
        f = np.dot(x[i], w) + b
        error = f - y[i]
        for j in range(n):
            gradient_w[j] += error * x[i][j]
        gradient_b += error
    gradient_w /= m
    gradient_b /= m
    return gradient_w, gradient_b
```

```python

# -----------------------------
# Gradient Descent
# -----------------------------
def gradient_descent(x, y, w, b, alpha, iterations):
    cost_history = []
    w = w.copy()
    for i in range(iterations):
        gradient_w, gradient_b = compute_gradient(x, y, w, b)
        w = w - alpha * gradient_w
        b = b - alpha * gradient_b
        cost = compute_cost(x, y, w, b)
        cost_history.append(cost)
    return w, b, cost_history

```

```python
# -----------------------------
# Run Gradient Descent
# -----------------------------
initial_w = np.zeros(X_train.shape[1])
initial_b = 0

alpha = 0.01  
iterations = 10000
w_final, b_final, hist = gradient_descent(
    X_train,
    y_train,
    initial_w,
    initial_b,
    alpha,
    iterations
)
print(f"w: {w_final}")
print(f"b: {b_final}")
print(f"Final Cost: {hist[-1]}")
```
