# Linear regreeison using tensor Flow

```python
import tensorflow as tf
import numpy as np
```

```python
#data
X_train = np.array([[1.0], [2.0]], dtype=np.float32)
Y_train = np.array([[300.0], [500.0]], dtype=np.float32)
```

```python
#create neuron
linear_layer = tf.keras.layers.Dense(units=1,activation='linear')
```

```python
#initiavlzie w (using no of inputs)
a1 = linear_layer(X_train[0].reshape(1,1))
```

```python
#set w & b
set_w = np.array([[200]])
set_b = np.array([100])
linear_layer.set_weights([set_w, set_b])
```

```python
#make prediction
prediction = linear_layer(X_train)
print(prediction)
```

# Logistic Neuron

```python
from tensorflow.keras import Sequential
```

```python
#data
X_train = np.array([0., 1, 2, 3, 4, 5], dtype=np.float32).reshape(-1,1)  # 2-D Matrix
Y_train = np.array([0,  0, 0, 1, 1, 1], dtype=np.float32).reshape(-1,1)  # 2-D Matrix
```

```python
#logistic regression Sequential model
model = Sequential([tf.keras.layers.Dense(1,input_dim=1,activation='sigmoid',name='L1')])
```

```python
#summary of modle
model.summary()
```

```python
#layer
logistic_layer =model.get_layer('L1')
```

```python
#intilaize 
a2 = model.predict(X_train[0].reshape(1,1))
```

```python
#set w & b
set_w = np.array([[2]])
set_b = np.array([-4.5])
logistic_layer.set_weights([set_w, set_b])
```

```python
#prediction 
prediction = model.predict(X_train)
print(prediction)
```

```python
for i in range(len(prediction)):
    
    if prediction[i] >= 0.5 :
        prediction[i] = 1
    else:
        prediction[i] = 0

print(prediction)
```
