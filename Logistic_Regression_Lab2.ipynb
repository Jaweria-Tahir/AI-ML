```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=2)
```

```python
#sigmoid functiion
def sigmoid(z):
    g = 1/(1+np.exp(z))
    return g
```

```python
#single value
z = 3
print(f'{sigmoid(z):.2f}')
```

```python
#array of values
z=np.array([2,3,4,5])
print(f'{sigmoid(z)}')
```

```python
arr = np.arange(-10,11)
ans=sigmoid(arr)
print(np.c_[arr,ans])
```

```python
fig, ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(arr, ans, c="b")

ax.set_title("Sigmoid Function")
ax.set_ylabel("sigmoid(z)")
ax.set_xlabel("z")

plt.show()
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression

X = [['Red'], ['Blue'], ['Green']]
y = [1, 0, 1]

encoder = OneHotEncoder(sparse_output=False)
X_encoded = encoder.fit_transform(X)

model = LogisticRegression()
model.fit(X_encoded, y)
```

```python
new_color=[['Blue']]
new_color_encoded = encoder.transform(new_color)
prediction = model.predict(new_color_encoded)
print(prediction)
```
