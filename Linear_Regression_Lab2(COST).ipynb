```python
import numpy as np
import matplotlib.pyplot as plt

x_train = np.array([1,2,3])
y_train = np.array([2,4,6])

plt.scatter(x_train,y_train,marker='X',c='r')
plt.xlabel('X-AXIS')
plt.ylabel('Y-AXIS')
plt.title('Graph')
plt.show()

def cost_function(x,y,w,b):
    m = x.shape[0]
    sum_cost=0
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w*x[i]+b
        cost = (f_wb[i]-y[i])**2
        sum_cost+=cost
        total_cost = (1/(2*m))*sum_cost
    return total_cost
b = 0 # b is fixed here
w_values = np.linspace(-1, 4, 100)
#create 100 evenly spaced values from -1 to 4
cost_array = []
for w in w_values: 
    cost = cost_function(x_train,y_train,w,b)
    cost_array.append(cost)

plt.plot(w_values,cost_array,c='r')
plt.xlabel('W')
plt.ylabel('J(w)')
plt.title('Cost Function Graph')
plt.show()


#minimum cost function is at w = 2
w = 2
f_wb = w * x_train + b

plt.scatter(x_train, y_train, marker='x', color='r', label='Actual')
plt.plot(x_train, f_wb, color='blue', label='Prediction')
plt.legend()
plt.show()
```
